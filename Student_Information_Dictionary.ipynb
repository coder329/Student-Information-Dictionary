# Student Information Dictionary 👨‍🎓📚

## Task 7 - Level 1 - Day 7

This project stores student information using a Python dictionary.

The program performs:
- Search operation
- Add operation
- Update operation
- Display of student records

```python
students = {
    "S001": {"name": "Rahul", "age": 20, "course": "AI & ML", "marks": 85},
    "S002": {"name": "Priya", "age": 21, "course": "Data Science", "marks": 91},
    "S003": {"name": "Aman", "age": 19, "course": "Computer Science", "marks": 78}
}

print(students)
```

## 1. Search Student

A unique student ID is used as the key to search for a student record.

```python
def search_student(student_id):
    if student_id in students:
        print("Student Found:")
        print("Student ID:", student_id)
        for key, value in students[student_id].items():
            print(f"{key.title()}: {value}")
    else:
        print("Student not found.")

search_student("S002")
```

## 2. Add Student

A new student can be added using a unique student ID.

```python
def add_student(student_id, name, age, course, marks):
    if student_id in students:
        print("Student ID already exists.")
    else:
        students[student_id] = {"name": name, "age": age, "course": course, "marks": marks}
        print("Student added successfully.")

add_student("S004", "Neha", 20, "Artificial Intelligence", 88)
print(students)
```

## 3. Update Student

Existing student information can be modified using the student ID and field name.

```python
def update_student(student_id, field, value):
    if student_id in students:
        students[student_id][field] = value
        print("Student information updated successfully.")
    else:
        print("Student not found.")

update_student("S001", "marks", 90)
print(students["S001"])
```

## 4. Display All Students

The following function displays all student records.

```python
def display_students():
    print("===== ALL STUDENT RECORDS =====")
    for student_id, information in students.items():
        print("\nStudent ID:", student_id)
        for key, value in information.items():
            print(f"{key.title()}: {value}")

display_students()
```

## 📊 Operations Performed

| Operation | Description |
|---|---|
| Search | Finds a student using the student ID |
| Add | Adds a new student record |
| Update | Modifies existing student information |
| Display | Shows all student records |

```python
search_student("S003")
add_student("S005", "Riya", 21, "Machine Learning", 94)
update_student("S002", "course", "Artificial Intelligence")
display_students()
```
